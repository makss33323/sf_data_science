```python
import pandas as pd

melb_data = pd.read_csv('data/melb_data_ps.csv', sep=',')
melb_data.head()
```

создаём рабочую копию для манипуляций

```python
melb_df = melb_data.copy()
melb_df.head()
```

```python
melb_df = melb_df.drop(['index', 'Coordinates'], axis=1)
melb_df.head()
```

```python
total_rooms = melb_df['Rooms'] + melb_df['Bedroom'] + melb_df['Bathroom']
display(total_rooms)

melb_df['MeanRoomsArea'] = melb_df['BuildingArea'] / total_rooms
display(melb_df['MeanRoomsArea'])

diff_area = melb_df['BuildingArea'] - melb_df['Landsize']
sum_area = melb_df['BuildingArea'] + melb_df['Landsize']
melb_df['AreaRatio'] = diff_area/sum_area
display(melb_df['AreaRatio'])
```

```python
melb_df['Date'] = pd.to_datetime(melb_df['Date'], dayfirst=True)
display(melb_df['Date'])
```

```python
import pandas as pd

# Предположим, что в melb_df есть столбец 'Date' с датами продаж
melb_df['Date'] = pd.to_datetime(melb_df['Date'])

# Создаём признак WeekdaySale (день недели)
melb_df['WeekdaySale'] = melb_df['Date'].dt.dayofweek

# Считаем количество объектов недвижимости, проданных в выходные (суббота и воскресенье)
weekend_count = melb_df[(melb_df['WeekdaySale'] == 5) | (melb_df['WeekdaySale'] == 6)].shape[0]

# Вывод результата
weekend_count
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Группируем данные по дням недели и считаем количество продаж
sales_by_weekday = melb_df.groupby('WeekdaySale').size()

# Преобразуем день недели из числового значения в название дня
sales_by_weekday.index = sales_by_weekday.index.map({0: 'Monday', 1: 'Tuesday', 2: 'Wednesday', 3: 'Thursday', 4: 'Friday', 5: 'Saturday', 6: 'Sunday'})

# Строим столбчатую диаграмму
plt.figure(figsize=(10, 6))
sns.barplot(x=sales_by_weekday.index, y=sales_by_weekday.values, palette='viridis')
plt.title('Продажи по дням недели')
plt.xlabel('День недели')
plt.ylabel('Количество продаж')
plt.xticks(rotation=45)
plt.show()
```
