Задан DataFrame countries_df, содержащий следующие столбцы: название страны, население (population) в миллионах человек и площадь страны (area) в квадратных километрах.

    'area': [133396, 9984670, 9826630, 17125191, 603628, 207600, 2724902]
})
Для каждой страны рассчитайте плотность населения (количество человек на квадратный километр).

Затем по полученным данным рассчитайте среднее по плотностям населения в указанных странах. Ответ округлите до сотых.

Плотность населения рассчитывается как количество человек, проживающих на территории отдельной страны, делённое на площадь этой страны. Обратите внимание, что население в таблице представлено в миллионах.

```python
import pandas as pd

# Заданный DataFrame
countries_df = pd.DataFrame({
    'country': ['Англия', 'Канада', 'США', 'Россия', 'Украина', 'Беларусь', 'Казахстан'],
    'population': [56.29, 38.05, 322.28, 146.24, 45.5, 9.5, 17.04],
    'area': [133396, 9984670, 9826630, 17125191, 603628, 207600, 2724902]
})

# Рассчитываем плотность населения
countries_df['population_density'] = (countries_df['population'] * 1e6) / countries_df['area']

# Рассчитываем среднюю плотность населения
average_density = countries_df['population_density'].mean()

# Округляем до сотых
average_density_rounded = round(average_density, 2)

print(f"Средняя плотность населения: {average_density_rounded} чел/кв.км")
```

```python
import pandas as pd

# Загрузка данных
melb_df = pd.read_csv('data/melb_data_ps.csv')

# Узнаем, сколько памяти занимает таблица до преобразования
memory_before = melb_df.memory_usage(deep=True).sum() / (1024 ** 2)  # Перевод байт в мегабайты

# Оставляем только 119 самых популярных пригородов, остальные заменяем на 'other'
top_119_suburbs = melb_df['Suburb'].value_counts().nlargest(119).index
melb_df['Suburb'] = melb_df['Suburb'].apply(lambda x: x if x in top_119_suburbs else 'other')

# Приводим столбец Suburb к категориальному типу
melb_df['Suburb'] = melb_df['Suburb'].astype('category')

# Узнаем, сколько памяти занимает таблица после преобразования
memory_after = melb_df.memory_usage(deep=True).sum() / (1024 ** 2)  # Перевод байт в мегабайты

# Вычисляем разницу в объеме занимаемой памяти
memory_difference = memory_before - memory_after

# Выводим результат, округлив до десятых
print(f"Разница в объеме занимаемой памяти: {memory_difference:.1f} МБ")
```
