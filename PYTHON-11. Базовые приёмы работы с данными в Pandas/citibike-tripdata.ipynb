Какой тип клиентов (столбец usertype) является преобладающим — Subscriber или Customer? В качестве ответа запишите долю клиентов преобладающего типа среди общего количества клиентов. Ответ округлите до сотых.

```python
import pandas as pd

# Загрузка данных из CSV-файла
data = pd.read_csv('data/citibike-tripdata.csv')

# Подсчет количества клиентов каждого типа
usertype_counts = data['usertype'].value_counts()

# Определение преобладающего типа клиентов
prevalent_usertype = usertype_counts.idxmax()

# Вычисление доли преобладающего типа среди общего количества клиентов
prevalent_usertype_ratio = usertype_counts.max() / usertype_counts.sum()

# Округление до сотых
prevalent_usertype_ratio_rounded = round(prevalent_usertype_ratio, 2)

print(f"Преобладающий тип клиентов: {prevalent_usertype}")
print(f"Доля клиентов преобладающего типа: {prevalent_usertype_ratio_rounded}")
```

```python
import pandas as pd

# Загрузка данных из CSV-файла
data = pd.read_csv('data/citibike-tripdata.csv')

# Преобразование столбцов starttime и stoptime в формат datetime
data['starttime'] = pd.to_datetime(data['starttime'])
data['stoptime'] = pd.to_datetime(data['stoptime'])

# Вычисление длительности поездки в минутах
data['trip duration'] = (data['stoptime'] - data['starttime']).dt.total_seconds() // 60

# Длительность поездки под индексом 3 в целых минутах
trip_duration_index_3 = data.loc[3, 'trip duration']
print(f'Длительность поездки под индексом 3: {trip_duration_index_3} минут')
```

```python
# Создание признака weekend
data['weekend'] = data['starttime'].dt.weekday.apply(lambda x: 1 if x >= 5 else 0)

# Подсчет количества поездок, начавшихся в выходные дни
num_weekend_trips = data['weekend'].sum()
print(f'Количество поездок, начавшихся в выходные дни: {num_weekend_trips}')
```

```python
import pandas as pd

# Загрузка данных из CSV-файла
data = pd.read_csv('data/citibike-tripdata.csv')

# Преобразование столбца starttime в формат datetime
data['starttime'] = pd.to_datetime(data['starttime'])

# Создание признака time_of_day
def get_time_of_day(hour):
    if 0 <= hour <= 6:
        return 'night'
    elif 6 < hour <= 12:
        return 'morning'
    elif 12 < hour <= 18:
        return 'day'
    elif 18 < hour <= 23:
        return 'evening'

data['time_of_day'] = data['starttime'].dt.hour.apply(get_time_of_day)

# Подсчет количества поездок по времени суток
num_day_trips = data[data['time_of_day'] == 'day'].shape[0]
num_night_trips = data[data['time_of_day'] == 'night'].shape[0]

# Вычисление отношения количества поездок, совершённых днём, к количеству поездок, совершённых ночью
ratio_day_to_night = round(num_day_trips / num_night_trips)
print(f'Во сколько раз количество поездок, совершённых днём, больше, чем количество поездок, совершённых ночью: {ratio_day_to_night}')
```
