```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

коробчатую диаграмму цен на квартиры (price_doc) в зависимости от уровня экологической обстановки в районе (ecology).

```python
# Определяем порядок категорий для ecology
ecology_order = ['excellent', 'good', 'satisfactory', 'poor', 'no data']  # Включи все категории из твоих данных

# Построение коробчатой диаграммы
plt.figure(figsize=(12, 6))
sns.boxplot(data=sber_data, x='ecology', y='price_doc', order=ecology_order)
plt.title('Коробчатая диаграмма цен на квартиры (price_doc) по уровню экологии')
plt.xlabel('Уровень экологии (ecology)')
plt.ylabel('Цена квартиры (price_doc)')

# Настройка формата оси Y
y_ticks = plt.gca().get_yticks()  # Получение текущих делений оси Y
plt.gca().set_yticklabels([f'{int(y/1e6)} млн' for y in y_ticks])  # Форматирование в миллионах

plt.xticks(rotation=45)  # Поворот подписей, если это нужно
plt.show()
```

диаграмму рассеяния, которая покажет, как цена на квартиру (price_doc) связана с расстоянием до центра Москвы (kremlin_km).

```python
import plotly.express as px

# Создание интерактивной диаграммы рассеяния
fig = px.scatter(sber_data, 
                 x='kremlin_km', 
                 y='price_doc', 
                 title='Связь между ценой квартиры и расстоянием до Кремля',
                 labels={'kremlin_km': 'Расстояние до Кремля (км)', 'price_doc': 'Цена квартиры'},
                 opacity=0.7)

# Отображение графика
fig.show()
```

Список столбцов с пропусками

```python
cols_null_percent = sber_data.isnull().mean() * 100
cols_with_null = cols_null_percent[cols_null_percent>0].sort_values(ascending=False)
display(cols_with_null)
```

Столбчатая диаграмма пропусков

```python
cols_with_null.plot(
    kind='bar',
    figsize=(10, 4),
    title='Распределение пропусков в данных'
);
```

Тепловая карта пропусков 

```python
colors = ['blue', 'yellow'] 
fig = plt.figure(figsize=(10, 4))
cols = cols_with_null.index
ax = sns.heatmap(
    sber_data[cols].isnull(),
    cmap=sns.color_palette(colors),
)
```
