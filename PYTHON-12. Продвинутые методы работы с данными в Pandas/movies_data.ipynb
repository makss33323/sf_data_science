userId — уникальный идентификатор пользователя, который выставил оценку;
movieId — уникальный идентификатор фильма;
rating — рейтинг фильма.

dates — таблица с датами выставления всех оценок.

movies — таблица с информацией о фильмах.

title — название фильма и год его выхода;
genres — жанры фильма.

```python
import pandas as pd
import os

def concat_user_files(path):
    # Список для хранения DataFrame-ов
    dfs = []

    # Итерация по всем файлам в указанной директории
    for filename in os.listdir(path):
        if filename.endswith('.csv'):
            filepath = os.path.join(path, filename)
            df = pd.read_csv(filepath)
            dfs.append(df)
    
    # Объединение всех DataFrame-ов в один
    combined_df = pd.concat(dfs)
    
    # Удаление дубликатов
    combined_df.drop_duplicates(inplace=True)
    
    # Обновление индексов
    combined_df.reset_index(drop=True, inplace=True)
    
    # Сортировка пользователей по числовой части user_id
    combined_df['user_id_numeric'] = combined_df['user_id'].str.extract('(\d+)').astype(int)
    combined_df.sort_values('user_id_numeric', inplace=True)
    combined_df.drop(columns='user_id_numeric', inplace=True)
    
    return combined_df

# Пример использования функции
path = './Root/users'
result_df = concat_user_files(path)
print(result_df)
```

```python
#библиотека для регулярных выражений
import re 
def get_year_release(arg):
    #находим все слова по шаблону "(DDDD)"
    candidates = re.findall(r'\(\d{4}\)', arg) 
    # проверяем число вхождений
    if len(candidates) > 0:
        #если число вхождений больше 0,
	#очищаем строку от знаков "(" и ")"
        year = candidates[0].replace('(', '')
        year = year.replace(')', '')
        return int(year)
    else:
        #если год не указан, возвращаем None
        return None
```
