```python
import pandas as pd

orders_df = pd.read_csv('data_orders_and_products/orders.csv', sep=';')
products_df = pd.read_csv('data_orders_and_products/products.csv', sep=';')
```

```python
# Переименование столбца для корректного объединения
orders_df = orders_df.rename(columns={'ID товара': 'Product_ID'})

# Объединение таблиц
orders_products = pd.merge(
    orders_df,
    products_df,
    on='Product_ID',
    how='inner'  # Только заказы, у которых есть соответствие в продуктах
)

# Вывод первых строк результирующей таблицы
print(orders_products.head())
```

```python
# Переименование столбца для корректного объединения
orders_df = orders_df.rename(columns={'ID товара': 'Product_ID'})

# Объединение таблиц
orders_products = pd.merge(
    orders_df,
    products_df,
    on='Product_ID',
    how='left'  # Все заказы, включая те, для которых нет информации о товарах
)

# Нахождение заказов без информации о товарах
missing_products_orders = orders_products[orders_products['Name'].isna()]

# Вывод идентификаторов таких заказов
order_ids = missing_products_orders['Order ID'].unique()

print("Идентификаторы заказов без информации о товарах:", order_ids)
```
